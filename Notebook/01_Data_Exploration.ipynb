# **Overview of All Datasets**

```python
datasets = {
    'Customers': customers,
    'Geolocation': geolocation,
    'Order Items': order_items,
    'Order Payments': order_payments,
    'Order Reviews': order_reviews,
    'Orders': orders,
    'Products': products,
    'Sellers': sellers,
    'Category Translation': category_translation
}

for name, df in datasets.items():
    print(name, ":", df.shape)
```

# **Finding_Data_Types**

```python
for name, df in datasets.items():
    print(f"\n{name}")
    summary = pd.DataFrame({
        'Column Name': df.columns,
        'Data Type': df.dtypes.values
    })
    display(summary)
```

# **Null_Values_Count**

```python
for name, df in datasets.items():
    print("\n==========", name, "==========")
    print(df.isnull().sum())
```

# **Missing_Values_count**

```python
for name, df in datasets.items():
    print("\n==========", name, "==========")

    missing_count = df.isnull().sum()

    missing_percentage = (df.isnull().sum() / len(df)) * 100

    missing_summary = pd.DataFrame({
        'Missing Values': missing_count,
        'Missing Percentage': missing_percentage
    })

    display(missing_summary[missing_summary['Missing Values'] > 0])
```

```python
for name, df in datasets.items():
    print("\n==========", name, "==========")

    missing_count = df.isnull().sum()

    missing_percentage = (df.isnull().sum() / len(df)) * 100

    missing_summary = pd.DataFrame({
        'Missing Values': missing_count,
        'Missing Percentage': missing_percentage
    })

    display(missing_summary[missing_summary['Missing Values'] > 0])
```

```python
missing_reports = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isnull().sum()

        missing_percentage = (missing_count / len(df)) * 100

        missing_reports.append({
            'Dataset': name,
            'Column': column,
            'Missing Values': missing_count,
            'Missing Percentage': round(missing_percentage, 2)
        })

missing_df = pd.DataFrame(missing_reports)

missing_df = missing_df[missing_df['Missing Values'] > 0]

missing_df = missing_df.sort_values(
    by='Missing Percentage',
    ascending=False
)

display(missing_df)
```

```python
missing_df.to_csv(
    os.path.join(processed_path, 'missing_values_report.csv'),
    index=False
)

print("Missing values report saved successfully!")
```

### **Duplicate_Values**

```python
for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    print(name, ":", duplicate_count, "duplicate rows")
```

```python
duplicate_report = []

for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    duplicate_report.append({
        'Dataset': name,
        'Duplicate Rows': duplicate_count,
        'Duplicate Percentage': round(
            (duplicate_count / len(df)) * 100, 2
        )
    })

duplicate_df = pd.DataFrame(duplicate_report)

duplicate_df.to_csv(
    os.path.join(processed_path, 'duplicate_records_report.csv'),
    index=False
)

display(duplicate_df)

print("Duplicate report saved successfully!")
```
